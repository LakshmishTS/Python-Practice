```python
# Sum of Numbers: Ask the user to enter numbers repeatedly. Stop when the user enters 0. Display the total sum.
total = 0
while True:
    print("Enter the number sum: ")
    num = int(input())
    total = total + num
    if num == 0:
        print("You entered 0 then it will break the sum loop")
        break

print("Sum of numbers is: ",total)
```

```python
# Guess the Number: Set a secret number as 25. Ask the user to guess it. Stop the loop when the correct number is entered.abs

secret_number = 25
while True:
    print("Enter your number for guess: ")
    num = int(input())
    if num == secret_number:
        print("You guessed correct number..!!!")
        break

print("Thank you")
```

```python
# ATM PIN: Ask the user to enter an ATM PIN. Stop when the correct PIN 5678 is entered and display Welcome to ATM.

correct_pin = 5678
while True:
    print("Enter your ATM pin:")
    num = int(input())
    if correct_pin != num:
        print("PIN IS WRONG...!!!Enter the correct pin")
        print()
    elif correct_pin == num:
        print("Welcome to ATM...Your pin is correct..!!!!!!")
        break

print("Thank you..!!")
```

```python
# Count Until 15: Print numbers from 1 to 100. Stop the loop when the number reaches 15.
num = 1

while num <= 100:
    print(num)

    if num == 15:
        break

    num = num + 1

print("Stopped at 15")
```

```python
# Shopping Bill: Ask the user to enter product prices repeatedly. Stop when the user enters 0. Display the total bill.abs

total = 0
while True:
    print("Enter the price of product: ")
    num = int(input())
    total = total + num
    if num == 0:
        print("You entered zero and your product lists is over")
        break

print(total)

print("Thank you")
```
