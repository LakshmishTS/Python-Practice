```python
# Print numbers from 1 to 50, but print only numbers divisible by 5.abs

for i in range(1,51):
    if i % 5 == 0:
        print(i)
```

```python
# Print numbers from 1 to 30. Skip numbers that are divisible by 3 using continue.abs

for i in range(1,30):
    if i % 3 == 0:
        continue
    print(i)
```

```python
# Print numbers from 1 to 50. Stop the loop when the number reaches 15 using break.

for i in range(1,51):
    if i == 16:
        break
    print(i)
```

```python
# Ask the user to enter 5 numbers. Calculate their total sum.

total = 0
for i in range(5):
    num = int(input("Enter the number: "))
    total = total + num
print("Total",total)
```

```python
# Ask the user to enter 10 numbers. Count how many are positive and how many are negative.abs

count_pos = 0
count_neg = 0
for i in range(1,11):
    num = int(input(f"Enter the number {i}: "))
    if num < 0:
        count_neg = count_neg + 1
    elif num > 0:
        count_pos = count_pos + 1
print()
print("Positive number: ",count_pos)
print("Negetive number: ",count_neg)
```

```python
# Ask the user to enter a password. Give the user 5 attempts to enter the correct password (python123).
# Stop immediately when the correct password is entered using break.abs

password = "python123"

for i in range(1, 6):
    passw = input(f"Enter your password (attempt {i}): ")
    if passw == password:
        print("Access granted!")
        break
    else:
        print("Wrong password.")
else:
    print("Too many failed attempts. Access denied.")
```
