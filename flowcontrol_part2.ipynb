```python
# Ask the user to enter numbers repeatedly. Stop when the user enters 0. Display the total sum.abs

total = 0
while True:
    print("Enter the number for sum: ")
    num = int(input())
    total = total + num
    if num == 0:
        break
print("Sum of total numbers is : ",total)
print("You entered 0 so it breaks the loop")
```

```python
# Ask the user to enter 10 numbers. Use continue to skip negative numbers. Display only positive numbers.abs

i = 1
while i <= 10:
    num = int(input(f"Enter the {i} number: "))
    i = i + 1
    if num < 0:
        continue
    print(num)
```

```python
# Set a secret number as 25. Ask the user to guess the number. Continue asking until the user enters 25. Use break to stop.abs

secret_number = 25
while True:
    print("Enter the number to guess:")
    num = int(input())
    if num != secret_number:
        continue
    else:
        print("Yes...!!!Your guess is correct..")
        break
```
